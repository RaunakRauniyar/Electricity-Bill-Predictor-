```python

# Electricity Bill Predictor
# Machine Learning Project

import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
import matplotlib.pyplot as plt

# 1. Read the dataset
df = pd.read_csv(
    "/content/household_power_consumption.txt",
    sep=";",
    na_values="?",
    low_memory=False
)

# 2. Select electricity consumption column
df["Global_active_power"] = pd.to_numeric(
    df["Global_active_power"], errors="coerce"
)

df = df.dropna(subset=["Global_active_power"])

# 3. Convert minute readings into daily electricity consumption
df["Daily_Units"] = df["Global_active_power"] / 60

daily = df.groupby("Date")["Daily_Units"].sum().reset_index()

# 4. Prepare data
daily["Day"] = range(1, len(daily) + 1)

X = daily[["Day"]]
y = daily["Daily_Units"]

# 5. Train and test the model
split = int(len(daily) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]
y_train = y.iloc[:split]
y_test = y.iloc[split:]

model = LinearRegression()
model.fit(X_train, y_train)

# 6. Check model accuracy
predictions = model.predict(X_test)

mae = mean_absolute_error(y_test, predictions)

print("Mean Absolute Error:", round(mae, 2), "units")

# 7. Predict future electricity consumption
next_day = len(daily) + 1

predicted_units = model.predict([[next_day]])[0]
predicted_units = max(0, predicted_units)

print("Predicted Electricity Consumption:",
      round(predicted_units, 2), "units")

# 8. Calculate electricity bill
rate = float(input("Enter electricity rate per unit (Rs.): "))

bill = predicted_units * rate

print("Estimated Electricity Bill: Rs.",
      round(bill, 2))

# 9. Show graph
plt.plot(y_test.values[:100], label="Actual Consumption")
plt.plot(predictions[:100], label="Predicted Consumption")

plt.xlabel("Day")
plt.ylabel("Electricity Units (kWh)")
plt.title("Electricity Consumption Prediction")
plt.legend()
plt.show()
```
